# Initial Data Exploration of ToN_IoT Dataset

## 1. Import Libraries

```python
import pandas as pd
```

## 2. Define Dataset Path and Instructions for User

```python
# Define the path to the dataset file
dataset_path = 'data/raw/ton_iot/Train_Test_Network.csv'

# IMPORTANT: User Action Required
# ----------------------------------
# Please download the ToN_IoT dataset. 
# You can typically find this by searching for 'ToN_IoT dataset' or 'UNSW ToN_IoT dataset'.
# Once downloaded, locate the 'Train_Test_datasets' folder (or similar).
# From that folder, copy the 'Train_Test_Network.csv' file into the `data/raw/ton_iot/` directory 
# in this project structure.
#
# The code below expects the file to be at: data/raw/ton_iot/Train_Test_Network.csv
```

## 3. Load the Dataset

```python
df = None
try:
    df = pd.read_csv(dataset_path)
    print(f"Successfully loaded {dataset_path}")
except FileNotFoundError:
    print(f"ERROR: Dataset file not found at {dataset_path}")
    print("Please make sure you have downloaded the ToN_IoT dataset and placed 'Train_Test_Network.csv' in the 'data/raw/ton_iot/' directory.")
except Exception as e:
    print(f"An error occurred while loading the dataset: {e}")
```

## 4. Basic DataFrame Exploration (if loaded)

```python
if df is not None:
    print("\n--- First 5 rows of the DataFrame ---")
    display(df.head())
    
    print("\n--- DataFrame Info ---")
    df.info()
    
    print("\n--- Descriptive Statistics ---")
    display(df.describe(include='all'))
    
    print("\n--- Missing Values per Column ---")
    missing_values = df.isnull().sum()
    print(missing_values[missing_values > 0] if not missing_values[missing_values > 0].empty else "No missing values found.")
    
    # Check for 'type' or 'label' column for class distribution
    # Assuming 'type' as per the prompt for now
    target_column = None
    if 'type' in df.columns:
        target_column = 'type'
    elif 'label' in df.columns: # Common alternative name
        target_column = 'label'
        
    if target_column:
        print(f"\n--- Value Counts for '{target_column}' column ---")
        display(df[target_column].value_counts())
    else:
        print("\n--- Target column ('type' or 'label') not found for class distribution analysis. ---")
else:
    print("\nDataFrame not loaded. Skipping further exploration steps.")
```
